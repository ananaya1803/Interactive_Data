```python

year_state = H1b_comp.groupby(
    ["Fiscal Year", "Petitioner State"]
)["NewEmpAppr_int"].sum().reset_index(name = "New Visa Approvals")

year_state_top10 = year_state[year_state["Petitioner State"].isin(state_newapprovals_top10.index.tolist())]

year_state_top10["State Name"] = year_state_top10["Petitioner State"].map(State_map)

fig6 = go.Figure()
for state in state_newapprovals_top10.index.tolist():
    state_data = year_state_top10[year_state_top10["Petitioner State"] == state]
    
    solid_data = state_data[state_data["Fiscal Year"] <= 2025]
    fig6.add_trace(go.Scatter(
        x=solid_data["Fiscal Year"],
        y=solid_data["New Visa Approvals"],
        mode="lines+markers",
        name=State_map[state],
        line=dict(color=trendcolor_map[state], dash="solid"),
        hovertemplate="<b>" + State_map[state] + "</b><br>Year: %{x}<br>Approvals: %{y:,}<extra></extra>"
    ))
    
    # Dotted line 2025-2026
    dotted_data = state_data[state_data["Fiscal Year"] >= 2025]
    fig6.add_trace(go.Scatter(
        x=dotted_data["Fiscal Year"],
        y=dotted_data["New Visa Approvals"],
        mode="lines+markers",
        name=State_map[state],
        line=dict(color=trendcolor_map[state], dash="dot"),
        showlegend=False,
        hovertemplate="<b>" + State_map[state] + "</b><br>Year: %{x}<br>Approvals: %{y:,} (incomplete)<extra></extra>"
    ))


fig6.update_layout(
    title=dict(
        text = "H1B Petition Trends by Top 10 States (2021-2026)",
        x = 0.5,
        xanchor = "center"),
    xaxis_title = "Fiscal Year",
    yaxis_title = "H1B Petitions",
    height = 500,
    margin={"r": 20, "t": 50, "l": 10, "b": 50},
    legend_title="State"
)

fig6.update_traces(
    line = dict(dash="dot"),
    selector=dict(name="2025")
)

fig6.show()

#source - https://claude.ai/chat/50981cae-65c2-43af-84f6-fcadafb66e60
```

Started building the Dash code with the skeletona and then filling it slowly and slowly.

```python
H1b_Dashboard_web = Dash()
H1b_Dashboard_web.layout = html.Div([
    html.H1("H1B Data Insight Dashboard"),
    dcc.Dropdown(
        options=[
            {'label' : "United States", "value" : "ALL"},
            {'label' : "California", "value" : "CA"},
            {"label" : "Texas", "value" : "TX"}
        ],
        id = "Dropdown",
        value = "ALL",
        clearable = False
    ),
    dcc.Graph(id = "Header_fig"),
    html.Div([
            dcc.Graph(id = "Left_fig", style = {"flex": 4}),
            dcc.Graph(id = "Right_fig", style = {"flex": 6})
        ],
        style = {"display" : "flex"})   
],
style={
    "backgroundColor": dashboard_colors["background"], 
    "minHeight": "100vh",   # makes it cover the full page height
    "padding": "40px",
    "boxSizing": "border-box"
})
#source - https://dash.plotly.com/layout

@callback(
Output("Header_fig", "figure"),
Output("Left_fig", "figure"),
Output("Right_fig", "figure"),
Input("Dropdown", "value")
)

def update_fig(value):
 if value == "CA":
    return fig10, fig3, fig8
 elif value == "TX":
    return fig11, fig4, fig9
 else:
    return fig1, fig5, fig6


if __name__== "__main__":
    H1b_Dashboard_web.run(debug = True)

    
```

##Limitations and Challenges  
1. One of the biggest challenge I faced was of narrowing my audience and data. The data was too vast to be able to visualize. I decided on narrowing it down to the Top 10 states, and Top 10 employers. I also narrowed it down to focus more on the two states and the industries with the highest number of approvals in them.  
2. I wanted to add different number of graphs for each value of the dropdown option. However, for the scope of this project, I decided to keep it symmetrical as it was easier to code with dash.
3. I also struggled with deciding on what information to convey, since this data consisted of so much information that could be relevant and at the same time, I also wanted to add more datasets to build more cohesive and focused dashboard. For this project, I decided to stick with this dataset and narrowed down the information that I wanted to convey with other ideas added as future work. 

 For future work, it would also help to focus on other variables like denials, continuation approval and other variables and other datasets could also be added to get a more comprehensive view. This dataset did not contain any demographic information. Adding that information (using census data or LCA data), I could get even more insight into H1B approvals and bring the information to a smaller number of audience.
