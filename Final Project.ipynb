```python
H1b = pd.read_csv("Employer Information.csv", encoding = "utf=16", sep = "\t")
H1b.head()
```

```python
H1b_2 = pd.read_csv("Employer Information 2.csv", encoding = "utf-16", sep="\t")
H1b_2.tail()
```

```python
H1b_comp = pd.concat([H1b_2, H1b], ignore_index= True)
H1b_comp.head()
```

```python
H1b_comp = H1b_comp.drop(columns=["Line by line"])
```

```python
H1b_comp.columns = H1b_comp.columns.str.strip()
```

```python
state_counts = H1b_comp.groupby("Petitioner State").count()['Employer (Petitioner) Name']
#state_counts_top10 = state_counts.nlargest(10)

fig1 = go.Figure(go.Choropleth(
    locations = state_counts.index,
    z = state_counts.values,
    text=state_counts.index.map(State_map),
    locationmode = "USA-states",
    colorscale = "Blues",
    colorbar_title = "H1B Petitions",
    hovertemplate="<b>%{text}</b><br>Petitions: %{z:,}<extra></extra>"
))

fig1.update_geos(
    visible = False, resolution = 110, scope = "usa",
    showcountries = True, countrycolor = "Black",
    showsubunits = True, 
    subunitcolor = "Black"
)

fig1.update_layout(
    title = "H1B Petitions by State (2021-2025)",
    height = 500,
    margin = {"r": 0, "t" : 50, "l" : 10, "b" : 50}
)

fig1.show()

# Source - https://claude.ai/chat/50981cae-65c2-43af-84f6-fcadafb66e60
```

```python
state_counts = H1b_comp.groupby("Petitioner State").count()['Employer (Petitioner) Name']
state_counts_top10 = state_counts.nlargest(10)
fig2 = go.Figure(
    data=[
    go.Pie(
        labels=state_counts_top10.index.map(State_map), 
        values=state_counts_top10.values,
        title="Petitioner States Percentage"
        )])

fig2.show()
```
