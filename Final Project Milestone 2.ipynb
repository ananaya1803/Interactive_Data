```python
import pandas as pd
from matplotlib import pyplot as plt
import plotly.graph_objects as go # or plotly.express as px
from plotly.subplots import make_subplots
import numpy as np
```

```python
H1b = pd.read_csv("Employer Information.csv", encoding = "utf=16", sep = "\t")
H1b.head()
```

```python
H1b.tail()
```

```python
H1b_2 = pd.read_csv("Employer Information 2.csv", encoding = "utf-16", sep="\t")
H1b_2.tail()
```

As I can see from the head() and tail(), the order of the years is descending. Its is the same for both the datasets, where one dataset startes from 2023, ending in 2021 and the other dataset starts from 2026, ending in 2024. When I combine these datasets, it would get jumbled. With the dates ordering 2023, 2022, 2021, 2026, 2025, 2024.

```python
H1b.columns = H1b.columns.str.strip()
H1b_2.columns = H1b_2.columns.str.strip()
```

```python
H1b["Fiscal Year"].value_counts()
```

```python
len(H1b)
```

```python
H1b_comp = pd.concat([H1b, H1b_2], ignore_index= True)
H1b_comp.head()
```

```python
H1b_comp.tail()
```

As expected, the dates have been jumbled up. I will resort this to make the dataset arranged as per the years in an ascending order.

```python
H1b_sort = H1b_comp.sort_values("Fiscal Year")
H1b_sort.tail()
```

I will be dropping the line by line column as it is just the indexing that pre-existed in the dataframe when it was downloaded. It is all jumbled up due to the resorting and it does not serve me any purpose

```python
H1b_sort = H1b_sort.drop(columns=["Line by line"])
```

```python
len(H1b_sort)
```

```python
H1b_sort.columns
```

```python
H1b_sort["Fiscal Year"].value_counts()
```

I will be making a simple graph with the Fiscal Year Information that I have.

```python
Fiscal_Yr = H1b_sort.groupby("Fiscal Year").count()["Employer (Petitioner) Name"]
Fiscal_Yr
```

```python
fig1, axs1 = plt.subplots()
axs1.bar(x = Fiscal_Yr.index, height = Fiscal_Yr.values)

plt.show()
```

Now I can see how many petitions were filed in every year from 2021 - 2026. This gives us an idea of H1B trends. As it can be seen, the number of petitions filed in 2026 is very low compared to other years. This could be due to applications being still in process or pending as the Fiscal Year 2026 is ongoing.
