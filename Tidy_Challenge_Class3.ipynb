for loop - [ x for x in]  
[len(x) for x in ]  
Using square brackets because of lists  
This is called list comprehension  
It iterates over data, applies a function to each element and returns a list to each output

```python
[len(x) for x in montero_tracks['available_markets']]
```

```python
['US' in x for x in montero_tracks['available_markets']]
```

```python
["Available" if "US" in x else "not available" for x in montero_tracks['available_markets']]
```

```python
montero_tracks["US"] = ["Available" if "US" in x else "not available" for x in montero_tracks['available_markets']]
montero_tracks.head()
```
