As we can see, several values for the variables "ConMediaUse" and "ImmiPhobia" are being repeated, like in ImmiPhobia, the value 1.0 comes 146 times and under "ConMediaUse", the value 1.000000 is repeated 138 times, because of which the scatterplot becomes overplotted and it makes it difficult to efficiently interpret the plot.  
As per Wilke (2019) we could potentially remedy this problem by using partial transparency. By making the plot points partially transparent, points which are overlapping become darker, making the shade of the points representative of the density of that particular plot point.  
However, in a data set like this, since the number of individual data points is large, partial transparency may not be sufficient and we another alternative method of improving this scatterplot could be to use a 2D histogram instead. Using gradient colors, we could represent repeated data points and singular data points. (Wilke, 2019)
