```python
year_state = H1b_comp.groupby(
    ["Fiscal Year", "Petitioner State"]
).count()["Employer (Petitioner) Name"].reset_index(name = "petitions")

year_state_top10 = year_state[year_state["Petitioner State"].isin(state_counts_top10)]

year_state_top10["State Name"] = year_state_top10["Petitioner State"].map(State_map)

fig5 = go.Figure()
for state in state_counts_top10.index.tolist():
    state_data = year_state_top10[year_state_top10["Petitioner State"] == state]
    fig5.add_trace(go.Scatter(
        x = state_data["Fiscal Year"],
        y = state_data["petitions"],
        mode = "lines+markers",
        name = State_map[state],
        hovertemplate="<b>" + State_map[state] + "</b><br>Year: %{x}<br>Petitions : %{y:, }<extra></extra>"        
    ))


fig5.update_layout(
    title = "H1B Petition Trends by Top 10 States (2021-2025)",
    xaxis_title = "Fiscal Year",
    yaxis_title = "H1B Petitions",
    height = 500,
    margin={"r": 20, "t": 50, "l": 10, "b": 50},
    xaxis=dict(dtick=1),                                 
    # shows every year on x axis
    legend_title="State"
)

fig5.show()
```

 I had to add .reset_index() to the groupby function because Petitioner State is stuck in Index and not acting as a regular column. Since we are using two variables to groupby, we'll use reset_index here so that Pandas treats both the variables as two distinct columns.
